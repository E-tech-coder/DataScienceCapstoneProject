**Further EDA**

Language analysis

```python
def detect_language(text):
    german_indicators = ['ü', 'ä', 'ö', 'ß', 'geschäftsführer', 'leiter', 'mitarbeiter']
    english_indicators = ['manager', 'director', 'senior', 'lead', 'engineer']

    text_lower = str(text).lower()
    german_score = sum(ind in text_lower for ind in german_indicators)
    english_score = sum(ind in text_lower for ind in english_indicators)

    if german_score > english_score:
        return 'German'
    elif english_score > german_score:
        return 'English'
    return 'Other'

df_linkedin_experience_annotated['language'] = df_linkedin_experience_annotated['position'].apply(detect_language)
print("Language distribution in job positions:")
print(df_linkedin_experience_annotated['language'].value_counts())

# Common words in positions
def get_top_words(series, n=20):
    all_words = ' '.join(series.str.lower()).split()
# Remove common words
    stopwords = {'und', 'der', 'die', 'das', 'and', 'the', 'of', 'for', 'in', 'at'}
    filtered_words = [w for w in all_words if w not in stopwords and len(w) > 2]
    return Counter(filtered_words).most_common(n)

print("  Top 20 words job positions (annotated data):")
top_words = get_top_words(df_linkedin_experience_annotated['position'])
for word, count in top_words[:20]:
    print(f"  {word}: {count}")
```

Carrer length, seniority progression and department changes

```python
# Calculate career length
career_lengths = df_linkedin_experience_annotated.groupby('person_id').apply(
    lambda x: len(x[x['status'].isin(['ACTIVE', 'INACTIVE'])])
)
print(f"---> Average number of positions per CV: {career_lengths.mean():.1f}")
print(f"   Min: {career_lengths.min()}, Max: {career_lengths.max()}")

# Seniority progression
seniority_order = {'Junior': 0, 'Professional': 1, 'Senior': 2, 'Lead': 3, 'Director': 4, 'Management': 5}
df_linkedin_experience_annotated['seniority_rank'] = df_linkedin_experience_annotated['seniority'].map(seniority_order)

progression_violations = 0
for person_id in df_linkedin_experience_annotated['person_id'].unique():
    person_data = df_linkedin_experience_annotated[df_linkedin_experience_annotated['person_id'] == person_id].sort_values('startDate')
    ranks = person_data['seniority_rank'].dropna()
    if len(ranks) > 1:
        if any(ranks.diff() < 0):
            progression_violations += 1

print(f"---> Career progression patterns:")
print(f"   Persons with seniority DECREASE: {progression_violations}")
print(f"    >> Validates assumption: seniority typically increases")

# Department changes
dept_changes = 0
for person_id in df_linkedin_experience_annotated['person_id'].unique():
    person_data = df_linkedin_experience_annotated[df_linkedin_experience_annotated['person_id'] == person_id]
    unique_depts = person_data['department'].nunique()
    if unique_depts > 1:
        dept_changes += 1

print(f"---> Department stability:")
print(f"   Persons who changed departments: {dept_changes} ({dept_changes/df_linkedin_experience_annotated['person_id'].nunique()*100:.1f}%)")
print(f"    >> Most people stay in same domain (validates assumption)")
```

Coverage analysis -> exact string matching

```python
print("Label Reference Datasets:")
print(f"   Department labels: {len(department_df)} job titles mapped to {department_df['label'].nunique()} classes")
print(f"   Seniority labels: {len(seniority_df)} job titles mapped to {seniority_df['label'].nunique()} classes")

# Check coverage of label files against annotated ACTIVE positions
active_positions = set(df_linkedin_experience_annotated['position'].str.lower())
dept_label_positions = set(department_df['text'].str.lower())
sen_label_positions = set(seniority_df['text'].str.lower())

# Exact matches
dept_exact_matches = len(active_positions.intersection(dept_label_positions))
sen_exact_matches = len(active_positions.intersection(sen_label_positions))

dept_coverage = dept_exact_matches / len(active_positions)
sen_coverage = sen_exact_matches / len(active_positions)

print(f"Coverage Analysis (exact string matching):")
print(f"   Department labels cover: {dept_coverage*100:.1f}% of ACTIVE positions ({dept_exact_matches}/{len(active_positions)})")
print(f"   Seniority labels cover: {sen_coverage*100:.1f}% of ACTIVE positions ({sen_exact_matches}/{len(active_positions)})")
print(f"\n   → {100-dept_coverage*100:.1f}% of positions need embedding for departments")
print(f"   → {100-sen_coverage*100:.1f}% of positions need embedding for seniority")
```

Exact string matching achieves only limited coverage (6–14%), demonstrating that real-world job titles exhibit high lexical variance. Therefore, semantic matching approaches such as embeddings are required to achieve sufficient coverage.
