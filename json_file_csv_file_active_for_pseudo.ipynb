```python
import json

PATH = "linkedin-cvs-not-annotated-to-be-predicted"

with open(PATH, "r", encoding="utf-8") as f:
    cvs = json.load(f)

print(type(cvs), len(cvs))
print(type(cvs[0]), len(cvs[0]))
print(cvs[0][0].keys())
```

```python
import pandas as pd

rows = []
for cv_id, jobs in enumerate(cvs):
    if not isinstance(jobs, list):
        continue
    for j in jobs:
        if not isinstance(j, dict):
            continue
        rows.append({
            "cv_id": cv_id,
            "organization": j.get("organization"),
            "position": j.get("position"),
            "status": j.get("status"),
            "startDate": j.get("startDate"),
            "endDate": j.get("endDate"),
            "linkedin": j.get("linkedin"),
        })

df_jobs = pd.DataFrame(rows)
df_jobs.shape, df_jobs.head(3)
```

```python
df_active = df_jobs[df_jobs["status"].astype(str).str.upper() == "ACTIVE"].copy()
df_active.shape
```

```python
ACTIVE_CSV = "linkedin_cvs_not_annotated_active_jobs.csv"
df_active.to_csv(ACTIVE_CSV, index=False)
ACTIVE_CSV
```
